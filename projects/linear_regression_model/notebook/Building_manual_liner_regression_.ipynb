```python
import pandas as pd
from sklearn import datasets
import matplotlib.pyplot as plt
```

```python
diabetes = datasets.load_diabetes()
```

```python
X = diabetes.data[:, 2] #USING BMI
Y = diabetes.target
```

```python
print(diabetes.feature_names)
```

```python
X.shape, Y.shape
```

```python

def gradient_descent(m_now, c_now, X, Y, L):
    m_gradient = 0
    c_gradient = 0
    n = len(X)

    for i in range(n):
        m_gradient += -(2/n) * X[i] * (Y[i] - (m_now * X[i] + c_now))
        c_gradient += -(2/n) * (Y[i] - (m_now * X[i] + c_now))

    m = m_now - m_gradient * L
    c = c_now - c_gradient * L
    return m, c

# Initialize parameters
m = 0
c = 0
L = 0.1
epochs = 10000

# Training loop
for i in range(epochs):
    m, c = gradient_descent(m, c, X, Y, L)

# Plot
plt.scatter(X, Y, color='black')
plt.plot(X, m * X + c, color='red')
plt.xlabel("BMI")
plt.ylabel("Disease Progression")
plt.show()
```
